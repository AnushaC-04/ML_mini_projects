```python
df=pd.read_csv('mrd_1000.csv')
```

```python
df=df.drop_duplicates()
```

```python
X=df.drop(columns='label',axis=1)
y=df['label']
```

```python
from sklearn.model_selection import train_test_split
Xtrain,Xtest,ytrain,ytest=train_test_split(X,y,train_size=0.8,random_state=42,stratify=y)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
pre=ColumnTransformer(
    transformers=[
        ('vector',TfidfVectorizer(max_features=5000,
             min_df=2,
             max_df=0.95),' text ')
    ]
)
```

```python
main_pipe=Pipeline(
    steps=[
        ('prep',pre),
        ('model',LogisticRegression(max_iter=1000,C=0.1))
    ]
)
```

```python
main_pipe.fit(Xtrain,ytrain)
```

```python
ypred_train=main_pipe.predict(Xtrain)
ypred_test=main_pipe.predict(Xtest)
```

```python
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
```

```python
print('accuracy_score: \n',accuracy_score(ytrain,ypred_train))
print('accuracy_score: \n',accuracy_score(ytest,ypred_test))
```

```python
print('confusion_matrix: \n',confusion_matrix(ytrain,ypred_train))
print('confusion_matrix: \n',confusion_matrix(ytest,ypred_test))
```

```python
print('Classification Report Train: ')
print(classification_report(ytrain,ypred_train))
print('Classification Report Test: ')
print(classification_report(ytest,ypred_test))
```

### Navie Bayes

```python
from sklearn.naive_bayes import MultinomialNB
nb_pipe=Pipeline(
    steps=[
        ('prep',pre),
        ('model',MultinomialNB())
    ]
)
```

```python
nb_pipe.fit(Xtrain,ytrain)
```

```python
ypred_train_NB = nb_pipe.predict(Xtrain)
ypred_test_NB = nb_pipe.predict(Xtest)
```

```python
print("Train Accuracy: ",accuracy_score(ytrain, ypred_train_NB))
print("Test Accuracy: ",accuracy_score(ytest, ypred_test_NB))
```

```python
print('Classification Report Train: ')
print(classification_report(ytrain,ypred_train_NB))
print('Classification Report Test: ')
print(classification_report(ytest, ypred_test_NB))
```

### SVC


```python
from sklearn.svm import LinearSVC
svm_pipe=Pipeline(
    steps=[
        ('prep',pre),
        ('model',LinearSVC(C=0.5))
    ]
)
```

```python
svm_pipe.fit(Xtrain,ytrain)
```

```python
ypred_train_SVC =svm_pipe.predict(Xtrain)
ypred_test_SVC =svm_pipe.predict(Xtest)
```

```python
print("Train Accuracy: ",accuracy_score(ytrain,ypred_train_SVC))
print("Test Accuracy: ",accuracy_score(ytest,ypred_test_SVC))
```

```python
print('Classification Report Train: ')
print(classification_report(ytrain,ypred_train_SVC))
print('Classification Report Test: ')
print(classification_report(ytest,ypred_test_SVC))
```

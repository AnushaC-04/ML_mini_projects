```python
df=df.drop(columns='Poster_Link',axis=1)
```

```python
df.shape
```

```python
df['Overview'].head()
```

```python
df['Content']=df['Series_Title'] +" " + df['Overview']
```

```python
df['Content']
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df['Certificate'].unique()
```

```python
df['Certificate']=df['Certificate'].fillna(df['Certificate'],'unknown')
```

```python
df['Gross']=df['Gross'].str.replace(',','',regex=False)
```

```python
df['Gross']=pd.to_numeric(df['Gross'],errors='coerce')
```

```python
df['Gross']
```

```python
df['Gross'] = df['Gross'].fillna(df['Gross'].median())
```

```python
df.info()
```

```python
X=df[['Content']]
y=df['Gross']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
Xtrain,Xtest,ytrain,ytest=train_test_split(X, y,train_size=0.8,random_state=42)
```

```python
model_V=TfidfVectorizer()
```

```python
Xtrain_vec = model_V.fit_transform(Xtrain['Content'])
Xtest_vec = model_V.transform(Xtest['Content'])
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
model=DecisionTreeRegressor(random_state=42)
```

```python
model.fit(Xtrain_vec,ytrain)
```

```python
ypred_train = model.predict(Xtrain_vec)
ypred_test = model.predict(Xtest_vec)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,root_mean_squared_error
```

```python
print('TRAIN: ')
print('r2_score: ',r2_score(ytrain,ypred_train))  
print('MAE: ',mean_absolute_error(ytrain,ypred_train))
print('RMSE: ',root_mean_squared_error(ytrain,ypred_train))
```

```python
print('TEST: ')
print('r2_score: ',r2_score(ytest,ypred_test))
print('MAE: ',mean_absolute_error(ytest,ypred_test))
print('RMSE: ',root_mean_squared_error(ytest,ypred_test))
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

```python

df=pd.DataFrame({
    'Age' :[18,21,45,32,np.nan,54,23,19,np.nan,44],
    'Salary':[25000,28000,np.nan,32000,np.nan,55000,np.nan,75000,65000,np.nan],
    'Likes' :['flower',np.nan,'birds',None,'flower',np.nan,'flower','birds',np.nan,'flower'],
    'Grade':['B','A',None,'D','C','Y',None,'A',np.nan,'Z']
    })
df
```

```python
Imputer=SimpleImputer(strategy='constant',fill_value=0)
Imputer.fit_transform(df[['Likes']])
```

```python
Imputer=SimpleImputer(strategy='constant',fill_value=0)  # it only check nan and replace it 
Imputer.fit_transform(df[['Likes','Grade']])
```

```python
Imputer=SimpleImputer(strategy='most_frequent')
Imputer.fit_transform(df[['Likes']])
```

```python
Imputer=SimpleImputer(strategy='mean')
Imputer.fit_transform(df[['Age','Salary']])
```

```python
Imputer=SimpleImputer(strategy='mean')
Imputer.fit_transform(df[['Age']])
```
